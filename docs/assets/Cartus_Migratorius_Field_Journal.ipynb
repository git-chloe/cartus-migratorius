```python

# ==========================================
# CHAPTER 2 — THE MYSTERY OF MISSING HABITATS
# 01 | Where Did the Coordinates Go?
# ==========================================

# Audit coordinate availability by complaint type
coordinate_audit = (
    carts.groupby("Description_EN")
    .agg(
        Total_Reports=("Has_Coordinates", "size"),
        With_Coordinates=("Has_Coordinates", "sum")
    )
)

# Calculate missing locations
coordinate_audit["Missing_Coordinates"] = (
    coordinate_audit["Total_Reports"]
    - coordinate_audit["With_Coordinates"]
)

coordinate_audit["Missing_Pct"] = (
    coordinate_audit["Missing_Coordinates"]
    / coordinate_audit["Total_Reports"] * 100
).round(1)

# Sort categories by missing coordinate counts
coordinate_audit = coordinate_audit.sort_values(
    "Missing_Coordinates",
    ascending=False
)

print("GEOGRAPHIC DISAPPEARANCE INVESTIGATION")
print("-" * 42)

print("Total shopping cart reports:", len(carts))
print(
    "Missing coordinates:",
    int((~carts["Has_Coordinates"]).sum())
)

print("\nCLASSIFICATIONS WITH MISSING COORDINATES")

display(
    coordinate_audit[
        coordinate_audit["Missing_Coordinates"] > 0
    ]
)

print("\nGEOGRAPHIC COVERAGE CHECK")

print(
    "Missing coordinates accounted for:",
    int(coordinate_audit["Missing_Coordinates"].sum())
)

print(
    "Classifications with complete coordinates:",
    int((coordinate_audit["Missing_Coordinates"] == 0).sum())
)

assert coordinate_audit["Missing_Coordinates"].sum() == 164

print("\n✓ Coordinate census reconciled.")
```

# Chapter 3: The Curious Case of the Municipal Time Machine ⏳

### When 2025's shopping carts travelled into 2026

During my initial investigation of Ottawa's municipal service records, I encountered a peculiar anomaly.

An unusually large number of service requests opened throughout 2025 appeared to have been closed during a single seven-day period in May 2026.

Had Ottawa experienced an extraordinary week of municipal productivity?

Had thousands of outstanding complaints suddenly been resolved?

Or was I observing something about the way municipal records were updated?

**Research question:** How widespread is this closure-date anomaly, and does it affect the shopping cart population?

Before attempting to identify repeated sightings or estimate reporting episodes, I must determine whether recorded closure dates can reliably tell me when an abandoned cart was actually removed.

*Field observation No. 006: Cartus migratorius has apparently developed an interest in time travel. The (fictional) field committee remains skeptical.*


```python

# ==========================================
# CHAPTER 3 — THE MUNICIPAL TIME MACHINE
# 01 | The Great Closure-Date Anomaly
# ==========================================

import pandas as pd

# Define the unusual seven-day period
spike_start = pd.Timestamp("2026-05-14")
spike_end = pd.Timestamp("2026-05-20")

# Convert closure dates to proper datetime values
# Invalid or unpublished dates become NaT
all_closed = pd.to_datetime(
    df["Closed Date"],
    errors="coerce"
)

cart_closed = pd.to_datetime(
    carts["Closed Date"],
    errors="coerce"
)

# Identify records in the unusual week
all_spike = all_closed.between(
    spike_start, spike_end
)

cart_spike = cart_closed.between(
    spike_start, spike_end
)

# Investigate all municipal requests
print("THE MUNICIPAL TIME MACHINE")
print("-" * 40)

print("ALL 311 SERVICE REQUESTS")
print(f"Total records: {len(df):,}")
print(f"Closed during unusual week: {all_spike.sum():,}")
print(f"Percentage: {all_spike.mean():.1%}")

print("\nSHOPPING CART REQUESTS")
print(f"Total cart reports: {len(carts):,}")
print(f"Closed during unusual week: {cart_spike.sum():,}")
print(f"Percentage: {cart_spike.mean():.1%}")

# Check the geocoded shopping cart subset
geocoded_carts = carts[
    carts["Has_Coordinates"]
].copy()

geocoded_closed = pd.to_datetime(
    geocoded_carts["Closed Date"],
    errors="coerce"
)

geocoded_spike = geocoded_closed.between(
    spike_start, spike_end
)

print("\nGEOCODED CART REPORTS")
print(f"Total: {len(geocoded_carts):,}")
print(f"Closed during unusual week: {geocoded_spike.sum():,}")
print(f"Percentage: {geocoded_spike.mean():.1%}")

# Show the daily concentration
print("\nCLOSURE RECORDS BY DAY")

daily_spike = (
    all_closed[all_spike]
    .value_counts()
    .sort_index()
)

display(daily_spike.rename("Service_Requests"))

# Validation against original field notes
assert all_spike.sum() == 121945, (
    "Closure count differs from original investigation."
)

print("\n✓ Original closure-date anomaly reproduced.")
```

```python

# ==========================================
# CHAPTER 3 — THE MUNICIPAL TIME MACHINE
# 02 | A Municipality-Wide Phenomenon?
# ==========================================

# Attach the previously calculated closure-date flag
service_investigation = df.assign(
    Closed_In_Unusual_Week=all_spike
)

# Examine the anomaly across municipal service types
service_audit = (
    service_investigation.groupby("Type")
    .agg(
        Total_Requests=("Closed_In_Unusual_Week", "size"),
        Closed_In_Week=("Closed_In_Unusual_Week", "sum")
    )
)

# Calculate how much of each service was affected
service_audit["Affected_Pct"] = (
    service_audit["Closed_In_Week"]
    / service_audit["Total_Requests"] * 100
).round(1)

# Sort by the number of unusual closure dates
service_audit = service_audit.sort_values(
    "Closed_In_Week",
    ascending=False
)

print("THE MUNICIPAL TIME MACHINE")
print("Investigation 02 — Municipal Services")
print("-" * 45)

display(service_audit)

# Check how widely the anomaly appears
affected_services = (
    service_audit["Closed_In_Week"] > 0
).sum()

print(
    "\nService types with affected records:",
    affected_services
)

# Confirm every affected record is accounted for
assert (
    service_audit["Closed_In_Week"].sum()
    == all_spike.sum()
)

print("\n✓ Service-level closure counts reconciled.")
```

```python

# ==========================================
# CHAPTER 3 — THE MUNICIPAL TIME MACHINE
# 03 | Did Reporting Channels Matter?
# ==========================================

# Prepare the closure-date and channel information
channel_investigation = df.assign(
    Closed_In_Unusual_Week=all_spike,
    Channel_Name=(
        df["Channel"]
        .astype("string")
        .str.strip()
        .fillna("Unspecified")
    )
)

# PART A: Investigate all municipal reporting channels
channel_audit = (
    channel_investigation
    .groupby("Channel_Name")
    .agg(
        Total_Requests=("Closed_In_Unusual_Week", "size"),
        Closed_In_Week=("Closed_In_Unusual_Week", "sum")
    )
)

channel_audit["Affected_Pct"] = (
    channel_audit["Closed_In_Week"]
    / channel_audit["Total_Requests"] * 100
).round(1)

print("THE MUNICIPAL TIME MACHINE")
print("Investigation 03 — Reporting Channels")
print("-" * 45)

print("\nALL MUNICIPAL REPORTING CHANNELS")
display(
    channel_audit.sort_values(
        "Total_Requests", ascending=False
    )
)

# PART B: Compare channels WITHIN major service types
# This avoids confusing differences between departments
# with differences caused by intake channels.

service_channel_audit = (
    channel_investigation
    .groupby(["Type", "Channel_Name"])
    .agg(
        Total_Requests=("Closed_In_Unusual_Week", "size"),
        Closed_In_Week=("Closed_In_Unusual_Week", "sum")
    )
    .reset_index()
)

service_channel_audit["Affected_Pct"] = (
    service_channel_audit["Closed_In_Week"]
    / service_channel_audit["Total_Requests"] * 100
).round(1)

selected_services = [
    "Roads and Transportation",
    "Water and the Environment",
    "Recreation and Culture",
    "Bylaw Services",
    "Garbage and Recycling"
]

print("\nREPORTING CHANNELS WITHIN SELECTED SERVICES")

display(
    service_channel_audit[
        service_channel_audit["Type"].isin(selected_services)
    ].sort_values(
        ["Type", "Total_Requests"],
        ascending=[True, False]
    )
)

# Confirm the channel breakdown reconciles
assert (
    channel_audit["Closed_In_Week"].sum()
    == all_spike.sum()
)

print("\n✓ Reporting-channel counts reconciled.")
```

```python

# ==========================================
# CHAPTER 3 — THE MUNICIPAL TIME MACHINE
# 04 | Did the Time Machine Reach Every Month?
# ==========================================

# Focus on Roads and Transportation
roads = df[
    df["Type"] == "Roads and Transportation"
].copy()

# Identify the unusual closure period
roads["Closed_In_Unusual_Week"] = all_spike.loc[
    roads.index
]

# Identify the month each request was opened
roads["Opened_Month"] = pd.to_datetime(
    roads["Opened Date"],
    errors="coerce"
).dt.to_period("M")

# Compare opening months against the closure anomaly
monthly_audit = (
    roads.groupby("Opened_Month")
    .agg(
        Total_Requests=("Closed_In_Unusual_Week", "size"),
        Closed_In_Week=("Closed_In_Unusual_Week", "sum")
    )
)

monthly_audit["Affected_Pct"] = (
    monthly_audit["Closed_In_Week"]
    / monthly_audit["Total_Requests"] * 100
).round(1)

print("THE MUNICIPAL TIME MACHINE")
print("Investigation 04 — Opening Months")
print("-" * 45)

display(monthly_audit)

# Verify all road requests are accounted for
assert monthly_audit["Total_Requests"].sum() == len(roads)
assert (
    monthly_audit["Closed_In_Week"].sum()
    == roads["Closed_In_Unusual_Week"].sum()
)

print("\n✓ Opening-month counts reconciled.")
```

```python

# ==========================================
# CHAPTER 4 — THE GREAT MIGRATION
# 04 | One Cart or Many?
# Part C: The 21-Day Social Gathering Rule
# ==========================================

import pandas as pd

# My exploratory reporting window
window_days = 21

episode_groups = []

# STEP 1: Examine each geographic cluster separately
for cluster_id, group in cart_clusters.groupby(
    "location_cluster"
):

    group = group.sort_values(
        ["Opened_Date", "Service Request ID"]
    ).copy()

    episode_number = 0
    episode_start = None
    episode_ids = []

    # STEP 2: Start a new episode when a report
    # falls more than 21 days after its first report
    for date in group["Opened_Date"]:

        if episode_start is None or (
            date - episode_start
        ).days > window_days:

            episode_number += 1
            episode_start = date

        episode_ids.append(
            f"{cluster_id}-{episode_number}"
        )

    group["episode_id"] = episode_ids
    episode_groups.append(group)

# STEP 3: Reassemble all reports
cart_episodes = pd.concat(
    episode_groups
).sort_index()

# STEP 4: Summarize the estimated episodes
spatial_episodes = (
    cart_episodes.groupby(
        ["location_cluster", "episode_id"]
    )
    .agg(
        Reports=("Service Request ID", "size"),
        Distinct_Addresses=("Address", "nunique"),
        First_Reported=("Opened_Date", "min"),
        Last_Reported=("Opened_Date", "max")
    )
    .reset_index()
)

print("THE 21-DAY SOCIAL GATHERING RULE")
print("-" * 42)

print("Original geocoded reports:", len(cart_episodes))
print(
    "Spatial clusters:",
    cart_episodes["location_cluster"].nunique()
)
print(
    "Estimated spatial-temporal episodes:",
    len(spatial_episodes)
)

print(
    "Episodes with multiple reports:",
    int((spatial_episodes["Reports"] > 1).sum())
)

print("\nOUR TWO CASE STUDIES")

display(
    spatial_episodes[
        spatial_episodes["location_cluster"].isin([3, 77])
    ]
)

# STEP 5: Reproducibility checks
assert len(cart_episodes) == 247
assert spatial_episodes["Reports"].sum() == 247

print("\n✓ Spatial-temporal census reconciled.")
```

```python

# ==========================================
# CHAPTER 4 — THE GREAT MIGRATION
# 04 | One Cart or Many?
# Part D: Address Matters!
# ==========================================

import pandas as pd

# STEP 1: Start with all 247 geocoded reports
address_episodes = cart_clusters.copy()

window_days = 21

# Create a place to store each report's episode ID
episode_ids = pd.Series(
    index=address_episodes.index,
    dtype="object"
)

# STEP 2: Investigate each recorded address separately
for address_id, (address, group) in enumerate(
    address_episodes.groupby("Address")
):

    group = group.sort_values(
        ["Opened_Date", "Service Request ID"]
    )

    episode_number = 0
    episode_start = None

    for idx, row in group.iterrows():

        report_date = row["Opened_Date"]

        # Start a new episode if more than 21 days
        # have passed since its FIRST report
        if (
            episode_start is None
            or (report_date - episode_start).days > window_days
        ):
            episode_number += 1
            episode_start = report_date

        # Give this report its estimated episode ID
        episode_ids.loc[idx] = (
            f"{address_id}-{episode_number}"
        )

# STEP 3: Attach episode IDs to the reports
address_episodes["address_episode"] = episode_ids

# STEP 4: Summarize the resulting episodes
address_episode_summary = (
    address_episodes.groupby("address_episode")
    .agg(
        Address=("Address", "first"),
        Reports=("Service Request ID", "size"),
        First_Reported=("Opened_Date", "min"),
        Last_Reported=("Opened_Date", "max")
    )
    .reset_index()
)

# Identify episodes with repeated reports
multi_report_episodes = address_episode_summary[
    address_episode_summary["Reports"] > 1
]

print("ADDRESS MATTERS!")
print("-" * 40)

print("Geocoded reports:", len(address_episodes))
print(
    "Distinct addresses:",
    address_episodes["Address"].nunique()
)
print(
    "Estimated address-based episodes:",
    len(address_episode_summary)
)
print(
    "Episodes with multiple reports:",
    len(multi_report_episodes)
)
print(
    "Additional reports within episodes:",
    len(address_episodes) - len(address_episode_summary)
)

# STEP 5: Revisit the familiar case studies
print("\nTHE WOODROFFE AND WALKLEY INVESTIGATIONS")

display(
    address_episode_summary[
        address_episode_summary["Address"].isin([
            "1701 Woodroffe Ave",
            "1245 Walkley Rd",
            "69 Lotta Ave"
        ])
    ].sort_values(["Address", "First_Reported"])
)

# STEP 6: Validate the new method
assert len(address_episodes) == 247
assert address_episodes["address_episode"].notna().all()
assert address_episode_summary["Reports"].sum() == 247
assert len(address_episode_summary) == 227

print("\n✓ Address-based episode census reconciled.")
```

```python

# ==========================================
# CHAPTER 4 — THE GREAT MIGRATION
# 04 | One Cart or Many?
# Part E: Testing Our Reporting Window
# ==========================================

import pandas as pd

# Repeat the same address-based episode calculation
# using alternative reporting windows.

def test_reporting_window(data, window_days):

    episode_sizes = []

    # Investigate each address independently
    for address, group in data.groupby("Address"):

        group = group.sort_values(
            ["Opened_Date", "Service Request ID"]
        )

        episode_start = None
        episode_size = 0

        for report_date in group["Opened_Date"]:

            # Begin another episode when the report
            # exceeds the window from the FIRST report
            if (
                episode_start is None
                or (report_date - episode_start).days > window_days
            ):

                # Preserve the previous episode's size
                if episode_size > 0:
                    episode_sizes.append(episode_size)

                episode_start = report_date
                episode_size = 0

            episode_size += 1

        # Preserve the final episode at each address
        if episode_size > 0:
            episode_sizes.append(episode_size)

    return {
        "Window_Days": window_days,
        "Estimated_Episodes": len(episode_sizes),
        "Multi_Report_Episodes": sum(
            size > 1 for size in episode_sizes
        ),
        "Additional_Reports": (
            len(data) - len(episode_sizes)
        )
    }


# Compare the four original assumptions
sensitivity_results = pd.DataFrame([
    test_reporting_window(address_episodes, days)
    for days in [7, 14, 21, 30]
])

print("THE GREAT REPORTING WINDOW EXPERIMENT")
print("-" * 45)

display(sensitivity_results)

# Confirm that every report remains accounted for
assert len(address_episodes) == 247

# Reproduce the selected 21-day result
assert (
    sensitivity_results.loc[
        sensitivity_results["Window_Days"] == 21,
        "Estimated_Episodes"
    ].iloc[0] == 227
)

print("\n✓ Reporting-window sensitivity test validated.")
```

### In Search of the Lone Wheel

Having documented the known habitats and recurring reporting locations of *Cartus migratorius*, the (fictional) field committee began to wonder about the less sociable members of the species.

- Which locations are the most geographically isolated?

### My methodology

I will examine the 214 distinct recorded addresses identified in Chapter 5.

Each address will be represented by the mean coordinates of its geocoded reports.

I will calculate the straight-line distance from each location to its nearest *different* recorded address.

**Important:** I am measuring isolation among reported locations, not distances travelled by individual shopping carts.

The absence of nearby reports does not establish the absence of shopping carts.

```python

# ==========================================
# CHAPTER 6 — THE GEOGRAPHY OF SOLITUDE
# 01 | The Nearest Neighbour
# ==========================================

import pandas as pd
import numpy as np

# STEP 1: Reuse the 214 distinct locations
# from Chapter 5's recurring habitat map

isolation_results = habitat_points.copy().reset_index(drop=True)

assert len(isolation_results) == 214

# STEP 2: Convert latitude and longitude
# from degrees into radians

lat = np.radians(
    isolation_results["Latitude"].to_numpy(dtype=float)
)

lon = np.radians(
    isolation_results["Longitude"].to_numpy(dtype=float)
)

# STEP 3: Calculate distances between every
# pair of distinct recorded locations

lat_difference = lat[:, None] - lat[None, :]
lon_difference = lon[:, None] - lon[None, :]

a = (
    np.sin(lat_difference / 2) ** 2
    + np.cos(lat[:, None])
    * np.cos(lat[None, :])
    * np.sin(lon_difference / 2) ** 2
)

earth_radius_km = 6371.0088

distance_matrix = (
    2 * earth_radius_km
    * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
)

# STEP 4: Exclude each location from being
# considered its own nearest neighbour

np.fill_diagonal(distance_matrix, np.inf)

# STEP 5: Identify the nearest OTHER location
nearest_indices = distance_matrix.argmin(axis=1)

nearest_distances = distance_matrix[
    np.arange(len(isolation_results)),
    nearest_indices
]

isolation_results["Nearest_Cart_Address"] = (
    isolation_results["Address"]
    .iloc[nearest_indices]
    .to_numpy()
)

isolation_results["Nearest_Cart_km"] = (
    nearest_distances.round(2)
)

# STEP 6: Rank the geographically isolated locations

lone_wheel_leaderboard = (
    isolation_results.sort_values(
        "Nearest_Cart_km",
        ascending=False
    )
)

print("THE GEOGRAPHY OF SOLITUDE")
print("-" * 42)

print("Distinct reported locations:", len(isolation_results))

print("\n🏆 THE LONE WHEEL AWARDS — TOP 10")

display(
    lone_wheel_leaderboard[
        [
            "Address",
            "Reports",
            "Nearest_Cart_Address",
            "Nearest_Cart_km"
        ]
    ].head(10)
)

# STEP 7: Examine the overall distance distribution

print("\nNEAREST-NEIGHBOUR DISTANCE SUMMARY")

display(
    isolation_results["Nearest_Cart_km"].describe().round(2)
)

# STEP 8: Validate the rankings

assert len(isolation_results) == 214
assert isolation_results["Nearest_Cart_Address"].notna().all()
assert (isolation_results["Nearest_Cart_km"] >= 0).all()
assert not (isolation_results["Address"] ==
            isolation_results["Nearest_Cart_Address"]).any()

print("\n✓ Nearest-neighbour investigation reconciled.")
print("✓ All 214 locations have been evaluated.")
```

```python

# ==========================================
# CHAPTER 6 — THE GEOGRAPHY OF SOLITUDE
# 04 | The Munster Effect
# ==========================================

import pandas as pd

# STEP 1: Preserve the full isolation dataset
all_habitats = habitat_isolation.copy()

# STEP 2: Create a comparison without Munster Road
without_munster = all_habitats[
    all_habitats["Address"] != "2890 Munster Rd"
].copy()

# Confirm I excluded exactly one Park location
munster_record = all_habitats[
    all_habitats["Address"] == "2890 Munster Rd"
]

assert len(munster_record) == 1
assert bool(munster_record["Park_Maintenance"].iloc[0])
assert len(without_munster) == 213

# STEP 3: Define a reusable summary function
def summarize_habitat_distances(data, scenario):

    summary = (
        data.groupby("Park_Maintenance")
        .agg(
            Locations=("Address", "size"),
            Median_km=("Nearest_Cart_km", "median"),
            Mean_km=("Nearest_Cart_km", "mean"),
            P90_km=(
                "Nearest_Cart_km",
                lambda x: x.quantile(0.90)
            )
        )
        .reset_index()
    )

    summary["Habitat"] = summary["Park_Maintenance"].map({
        True: "Park Maintenance",
        False: "Other Locations"
    })

    summary["Scenario"] = scenario

    return summary[[
        "Scenario", "Habitat", "Locations",
        "Median_km", "Mean_km", "P90_km"
    ]]


# STEP 4: Compare both scenarios
munster_sensitivity = pd.concat([
    summarize_habitat_distances(
        all_habitats, "All 214 locations"
    ),
    summarize_habitat_distances(
        without_munster, "Without Munster"
    )
], ignore_index=True)

print("THE MUNSTER EFFECT")
print("-" * 45)

display(munster_sensitivity.round(2))

# STEP 5: Calculate Munster's influence
park_before = all_habitats[
    all_habitats["Park_Maintenance"]
]["Nearest_Cart_km"]

park_after = without_munster[
    without_munster["Park_Maintenance"]
]["Nearest_Cart_km"]

print("\nPARK MAINTENANCE — CHANGE IN MEAN")

print(f"Original mean: {park_before.mean():.2f} km")
print(f"Without Munster: {park_after.mean():.2f} km")

print(
    "Percentage reduction:",
    f"{(1 - park_after.mean() / park_before.mean()) * 100:.1f}%"
)

# STEP 6: Validate
assert len(all_habitats) == 214
assert len(without_munster) == 213
assert len(park_before) == 29
assert len(park_after) == 28

print("\n✓ Outlier sensitivity test reconciled.")
print("✓ Original geographic measurements preserved.")
```

```python

# ==========================================
# CHAPTER 6 — THE GEOGRAPHY OF SOLITUDE
# 05 | Beyond the One-Kilometre Frontier
# ==========================================

import pandas as pd

# STEP 1: Compare two geographic scenarios
scenarios = [
    ("All Locations", habitat_isolation),
    ("Without Munster", without_munster)
]

# The two exploratory isolation thresholds
thresholds = [1.0, 1.5]

results = []

# STEP 2: Test each scenario and threshold
for scenario_name, dataset in scenarios:

    for threshold in thresholds:

        for is_park, habitat_name in [
            (True, "Park Maintenance"),
            (False, "Other Locations")
        ]:

            group = dataset[
                dataset["Park_Maintenance"] == is_park
            ]

            # Count locations beyond the threshold
            beyond = (
                group["Nearest_Cart_km"] > threshold
            ).sum()

            results.append({
                "Scenario": scenario_name,
                "Threshold_km": threshold,
                "Habitat": habitat_name,
                "Total_Locations": len(group),
                "Beyond_Threshold": int(beyond),
                "Percentage_Beyond": round(
                    beyond / len(group) * 100, 1
                )
            })

# STEP 3: Assemble the comparison
distance_thresholds = pd.DataFrame(results)

print("BEYOND THE ONE-KILOMETRE FRONTIER")
print("-" * 45)

display(distance_thresholds)

# STEP 4: Confirm the geographic populations
assert len(habitat_isolation) == 214
assert len(without_munster) == 213
assert len(distance_thresholds) == 8

print("\n✓ Isolation threshold experiment reconciled.")
```

```python

# ==========================================
# CHAPTER 7 — THE GOLDEN WHEEL EXPEDITION
# 06 | The Great Retail Doppelgänger Investigation
# ==========================================

import pandas as pd
import numpy as np

# STEP 1: Create the provisional storefront population
# Keep both core and extended candidates for now.

eligible_statuses = [
    "Core candidate - verify location",
    "Extended candidate - cart use uncertain"
]

storefront_review = (
    retailer_review[
        retailer_review["Review_Status"].isin(
            eligible_statuses
        )
    ]
    .copy()
    .rename_axis("Source_Index")
    .reset_index()
)

print("THE GREAT RETAIL DOPPELGÄNGER INVESTIGATION")
print("-" * 50)
print("Original applicants:", len(retailer_review))
print("Provisional storefront records:", len(storefront_review))

print("\nCANDIDATES BY BRAND")
display(
    storefront_review["Brand"]
    .value_counts()
)

# STEP 2: Identify same-brand locations within
# 120 metres of one another.
#
# This is a screening distance, not proof that
# two records describe the same physical store.

duplicate_radius_m = 120
earth_radius_m = 6371008.8

candidate_pairs = []

for brand, group in storefront_review.groupby("Brand"):

    group = group.reset_index(drop=True)

    if len(group) < 2:
        continue

    # Convert all coordinates into radians
    coordinates = np.radians(
        group[["Latitude", "Longitude"]]
        .to_numpy(dtype=float)
    )

    lat = coordinates[:, 0]
    lon = coordinates[:, 1]

    # Calculate pairwise distances using haversine
    dlat = lat[:, None] - lat[None, :]
    dlon = lon[:, None] - lon[None, :]

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat[:, None])
        * np.cos(lat[None, :])
        * np.sin(dlon / 2) ** 2
    )

    distances = (
        2 * earth_radius_m
        * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
    )

    # Examine each pair once, excluding self-pairs
    rows, cols = np.where(
        np.triu(
            distances <= duplicate_radius_m,
            k=1
        )
    )

    for i, j in zip(rows, cols):

        candidate_pairs.append({
            "Brand": brand,
            "Store_1": group.loc[i, "Name"],
            "Store_2": group.loc[j, "Name"],
            "Index_1": group.loc[i, "Source_Index"],
            "Index_2": group.loc[j, "Source_Index"],
            "Category_1": group.loc[i, "Category"],
            "Category_2": group.loc[j, "Category"],
            "Distance_m": round(
                distances[i, j], 1
            )
        })

# STEP 3: Assemble the possible duplicate pairs

possible_store_duplicates = pd.DataFrame(
    candidate_pairs,
    columns=[
        "Brand", "Store_1", "Store_2",
        "Index_1", "Index_2",
        "Category_1", "Category_2",
        "Distance_m"
    ]
).sort_values(
    ["Distance_m", "Brand"]
).reset_index(drop=True)

print("\nPOSSIBLE DUPLICATE STORE PAIRS")
print("Distance threshold:", duplicate_radius_m, "metres")
print("Pairs requiring review:", len(possible_store_duplicates))

display(possible_store_duplicates)

# STEP 4: Identify which brands need attention

print("\nPOSSIBLE DUPLICATE PAIRS BY BRAND")

display(
    possible_store_duplicates["Brand"]
    .value_counts()
)

# STEP 5: Validate — no records removed
assert len(retailer_review) == 795
assert len(storefront_review) == 307
assert storefront_review["Source_Index"].is_unique

print("\n✓ Storefront screening population reconciled.")
print("✓ Nearby same-brand records identified.")
print("✓ No records removed or merged.")
```

```python

# ==========================================
# CHAPTER 7 — THE GOLDEN WHEEL EXPEDITION
# 13 | The Final Four Background Checks
# ==========================================

import pandas as pd
import numpy as np

# STEP 1: Preserve the four unusual core candidates
final_four = remaining_category_review.copy()

assert len(final_four) == 4

earth_radius_km = 6371.0088

# STEP 2: Create a reusable distance function
# Calculate straight-line distance from one
# location to a collection of other locations.

def distances_from_location(lat, lon, other_locations):

    lat1 = np.radians(float(lat))
    lon1 = np.radians(float(lon))

    lat2 = np.radians(
        other_locations["Latitude"].to_numpy(dtype=float)
    )
    lon2 = np.radians(
        other_locations["Longitude"].to_numpy(dtype=float)
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return (
        2 * earth_radius_km
        * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
    )


# STEP 3: Compare each candidate against
# other stores belonging to the same brand.

background_checks = []

for _, candidate in final_four.iterrows():

    same_brand = core_stage2[
        (core_stage2["Brand"] == candidate["Brand"])
        &
        (
            core_stage2["Source_Index"]
            != candidate["Source_Index"]
        )
    ].copy()

    if len(same_brand) > 0:

        distances = distances_from_location(
            candidate["Latitude"],
            candidate["Longitude"],
            same_brand
        )

        nearest_position = distances.argmin()
        nearest_store = same_brand.iloc[nearest_position]

        nearest_brand_name = nearest_store["Name"]
        nearest_brand_km = distances[nearest_position]

    else:
        nearest_brand_name = None
        nearest_brand_km = np.nan

    # STEP 4: Measure proximity to the nearest
    # geographically identifiable cart location.

    cart_distances = distances_from_location(
        candidate["Latitude"],
        candidate["Longitude"],
        habitat_points
    )

    nearest_cart_km = cart_distances.min()

    background_checks.append({
        "Source_Index": candidate["Source_Index"],
        "Name": candidate["Name"],
        "Brand": candidate["Brand"],
        "Category": candidate["Category"],
        "Latitude": candidate["Latitude"],
        "Longitude": candidate["Longitude"],
        "Nearest_Same_Brand": nearest_brand_name,
        "Same_Brand_km": round(nearest_brand_km, 2),
        "Nearest_Cart_km": round(nearest_cart_km, 2)
    })

# STEP 5: Assemble the results

final_four_audit = pd.DataFrame(background_checks)

print("THE FINAL FOUR BACKGROUND CHECKS")
print("-" * 48)

display(final_four_audit)

# STEP 6: Compare the reconstructed core
# with the count from my Original Field Notes.

print("\nCORE RETAILER CENSUS COMPARISON")
print("-" * 48)

print("Original Field Notes core count: 144")
print("Current provisional core count:", len(core_stage2))
print(
    "Difference:",
    len(core_stage2) - 144
)

print("\nCANDIDATE STORE COUNTS BY BRAND")

display(
    core_stage2["Brand"]
    .value_counts()
    .sort_index()
)

# STEP 7: Validate

assert len(final_four_audit) == 4
assert len(core_stage2) == 151
assert len(retailer_stage2) == 273
assert final_four_audit["Nearest_Cart_km"].notna().all()

print("\n✓ Four candidates reviewed geographically.")
print("✓ Original records unchanged.")
print("✓ No additional exclusions made.")
```

```python

# ==========================================
# CHAPTER 7 — THE GOLDEN WHEEL EXPEDITION
# 15 | The Golden Measuring Tape
# ==========================================

import pandas as pd
import numpy as np

# STEP 1: Reuse the cleaned study populations
cart_locations = habitat_points.copy()

core_candidates = core_stage3.copy()
extended_candidates = extended_stage3.copy()

assert len(cart_locations) == 214
assert len(core_candidates) == 150
assert len(extended_candidates) == 272

# STEP 2: Build a reusable nearest-retailer function
# Distances use the haversine formula and are
# measured in kilometres, not road distances.

def find_nearest_retailers(locations, retailers):

    cart_lat = np.radians(
        locations["Latitude"].to_numpy(dtype=float)
    )[:, None]

    cart_lon = np.radians(
        locations["Longitude"].to_numpy(dtype=float)
    )[:, None]

    store_lat = np.radians(
        retailers["Latitude"].to_numpy(dtype=float)
    )[None, :]

    store_lon = np.radians(
        retailers["Longitude"].to_numpy(dtype=float)
    )[None, :]

    dlat = cart_lat - store_lat
    dlon = cart_lon - store_lon

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(cart_lat)
        * np.cos(store_lat)
        * np.sin(dlon / 2) ** 2
    )

    distances = (
        2 * 6371.0088
        * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
    )

    # Find the closest retailer for every
    # recorded shopping cart location
    nearest_positions = distances.argmin(axis=1)

    nearest_stores = retailers.iloc[
        nearest_positions
    ].reset_index(drop=True)

    results = locations[
        ["Address", "Reports", "Latitude", "Longitude"]
    ].reset_index(drop=True).copy()

    results["Nearest_Store"] = (
        nearest_stores["Name"].to_numpy()
    )

    results["Nearest_Brand"] = (
        nearest_stores["Brand"].to_numpy()
    )

    results["Nearest_Store_ID"] = (
        nearest_stores["Source_Index"].to_numpy()
    )

    # Retain full numerical precision
    results["Distance_km"] = distances[
        np.arange(len(locations)),
        nearest_positions
    ]

    return results


# STEP 3: Calculate both retailer scenarios
core_distances = find_nearest_retailers(
    cart_locations, core_candidates
)

extended_distances = find_nearest_retailers(
    cart_locations, extended_candidates
)

# STEP 4: Compare their results address by address
golden_comparison = core_distances[
    ["Address", "Reports", "Latitude", "Longitude"]
].copy()

golden_comparison["Core_km"] = (
    core_distances["Distance_km"]
)

golden_comparison["Core_Store"] = (
    core_distances["Nearest_Store"]
)

golden_comparison["Core_Store_ID"] = (
    core_distances["Nearest_Store_ID"]
)

golden_comparison["Extended_km"] = (
    extended_distances["Distance_km"]
)

golden_comparison["Extended_Store"] = (
    extended_distances["Nearest_Store"]
)

golden_comparison["Extended_Store_ID"] = (
    extended_distances["Nearest_Store_ID"]
)

golden_comparison["Difference_km"] = (
    golden_comparison["Core_km"]
    - golden_comparison["Extended_km"]
)

# STEP 5: Display both leaderboards
print("THE GOLDEN MEASURING TAPE")
print("-" * 45)

print("Shopping cart reporting locations:",
      len(cart_locations))
print("Core retailer candidates:",
      len(core_candidates))
print("Extended retailer candidates:",
      len(extended_candidates))

print("\n🏆 TOP 10 — CORE RETAILER SCENARIO")

display(
    golden_comparison.nlargest(10, "Core_km")[
        ["Address", "Core_km", "Core_Store",
         "Core_Store_ID"]
    ].round(2)
)

print("\n🏆 TOP 10 — EXTENDED RETAILER SCENARIO")

display(
    golden_comparison.nlargest(10, "Extended_km")[
        ["Address", "Extended_km", "Extended_Store",
         "Extended_Store_ID"]
    ].round(2)
)

# STEP 6: Find where retailer assumptions matter most
print("\nLARGEST DISTANCE CHANGES BETWEEN SCENARIOS")

display(
    golden_comparison.nlargest(10, "Difference_km")[
        ["Address", "Core_km", "Extended_km",
         "Difference_km", "Extended_Store"]
    ].round(2)
)

# STEP 7: Revisit the familiar rural contestant
print("\nTHE MUNSTER ROAD FILE")

display(
    golden_comparison[
        golden_comparison["Address"]
        == "2890 Munster Rd"
    ].round(2)
)

# STEP 8: Validate
assert len(golden_comparison) == 214
assert len(core_distances) == 214
assert len(extended_distances) == 214

# The extended scenario contains every core store.
# Adding candidates cannot increase nearest distance.
assert (
    golden_comparison["Extended_km"]
    <= golden_comparison["Core_km"] + 1e-8
).all()

print("\n✓ Both retailer scenarios reconciled.")
print("✓ All 214 reported locations evaluated.")
print("✓ No retailer origins assumed or assigned.")
```

```python

# ==========================================
# CHAPTER 8 — THE GREAT CART ESCAPE
# 03 | The Double Isolation Club
# ==========================================

import pandas as pd
import numpy as np

print("THE DOUBLE ISOLATION CLUB")
print("-" * 52)

# STEP 1: Retrieve the 214 distinct cart
# reporting locations from Chapter 6.

neighbour_points = habitat_points.copy()

assert len(neighbour_points) == 214
assert neighbour_points["Address"].is_unique

# STEP 2: Recalculate the distance to the
# nearest OTHER recorded cart location.
#
# This independently reproduces the Chapter 6
# nearest-neighbour measurement.

latitude = np.radians(
    neighbour_points["Latitude"].to_numpy(dtype=float)
)

longitude = np.radians(
    neighbour_points["Longitude"].to_numpy(dtype=float)
)

dlat = latitude[:, None] - latitude[None, :]
dlon = longitude[:, None] - longitude[None, :]

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(latitude[:, None])
    * np.cos(latitude[None, :])
    * np.sin(dlon / 2) ** 2
)

distance_matrix = (
    2 * 6371.0088
    * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
)

# A location cannot be its own nearest neighbour.
np.fill_diagonal(distance_matrix, np.inf)

nearest_positions = distance_matrix.argmin(axis=1)

neighbour_results = pd.DataFrame({
    "Address": neighbour_points["Address"].values,
    "Nearest_Cart_Address": (
        neighbour_points.iloc[nearest_positions]
        ["Address"].values
    ),
    "Cart_to_Cart_km": distance_matrix[
        np.arange(len(neighbour_points)),
        nearest_positions
    ]
})

# STEP 3: Combine the two measures of isolation

double_solitude = park_escape.merge(
    neighbour_results,
    on="Address",
    how="left",
    validate="one_to_one"
)

assert len(double_solitude) == 214
assert double_solitude["Cart_to_Cart_km"].notna().all()

# STEP 4: Define the exploratory threshold:
# more than 1 km under EACH measurement.
#
# These are geographic categories, not evidence
# of an actual shopping cart journey.

double_solitude["Far_From_Carts"] = (
    double_solitude["Cart_to_Cart_km"] > 1.0
)

double_solitude["Far_From_Retail"] = (
    double_solitude["Core_km"] > 1.0
)

double_solitude["Double_Isolated"] = (
    double_solitude["Far_From_Carts"]
    & double_solitude["Far_From_Retail"]
)

# STEP 5: Examine all four possible combinations

print("\nTHE FOUR GEOGRAPHIC PERSONALITIES")

isolation_table = pd.crosstab(
    double_solitude["Far_From_Carts"],
    double_solitude["Far_From_Retail"]
)

isolation_table.index = [
    "Near another cart report",
    "Far from other cart reports"
]

isolation_table.columns = [
    "Within 1 km of retailer",
    "Over 1 km from retailer"
]

display(isolation_table)

# STEP 6: Compare the two service populations

group_summary = (
    double_solitude
    .groupby("Habitat_Group")
    .agg(
        Locations=("Address", "size"),
        Far_From_Carts=("Far_From_Carts", "sum"),
        Far_From_Retail=("Far_From_Retail", "sum"),
        Double_Isolated=("Double_Isolated", "sum")
    )
)

group_summary["Double_Isolated_Pct"] = (
    group_summary["Double_Isolated"]
    / group_summary["Locations"]
    * 100
)

print("\nWHO JOINS THE DOUBLE ISOLATION CLUB?")

display(group_summary.round(1))

# STEP 7: Meet the members

print("\nDOUBLE ISOLATION CLUB — MEMBERS")
print("Both distances exceed 1 km")

club_members = (
    double_solitude[
        double_solitude["Double_Isolated"]
    ]
    .sort_values(
        "Cart_to_Cart_km",
        ascending=False
    )
)

display(
    club_members[
        [
            "Address",
            "Habitat_Group",
            "Cart_to_Cart_km",
            "Core_km",
            "Extended_km",
            "Nearest_Cart_Address",
            "Core_Store"
        ]
    ].round(2)
)

# STEP 8: Explore whether the two distances
# tend to increase together.
#
# Spearman measures rank association.
# It does not establish causation.

with_munster = double_solitude[
    ["Cart_to_Cart_km", "Core_km"]
].corr(method="spearman").iloc[0, 1]

without_munster = double_solitude[
    double_solitude["Address"] != "2890 Munster Rd"
][
    ["Cart_to_Cart_km", "Core_km"]
].corr(method="spearman").iloc[0, 1]

print("\nDO THE TWO TYPES OF ISOLATION TRACK TOGETHER?")

print("Spearman correlation — all locations:",
      round(with_munster, 3))

print("Spearman correlation — without Munster:",
      round(without_munster, 3))

# STEP 9: Validate against the earlier study

munster_check = double_solitude[
    double_solitude["Address"] == "2890 Munster Rd"
].iloc[0]

assert abs(
    munster_check["Cart_to_Cart_km"] - 11.04
) < 0.02

assert abs(
    munster_check["Core_km"] - 9.69
) < 0.02

assert len(double_solitude) == 214
assert isolation_table.to_numpy().sum() == 214

print("\n✓ All 214 reporting locations classified.")
print("✓ Cart-to-cart distances reconstructed.")
print("✓ Retailer isolation compared.")
print("✓ Munster's double championship reconciled.")
print("✓ No original records modified.")
```

```python

# ==========================================
# CHAPTER 8 — THE GREAT CART ESCAPE
# 04 | The Membership Committee
# ==========================================

import pandas as pd
import numpy as np

print("THE DOUBLE ISOLATION MEMBERSHIP COMMITTEE")
print("-" * 54)

# STEP 1: Define alternative club admission rules
# A location must exceed BOTH distances.

membership_thresholds = [
    0.75,
    1.00,
    1.25,
    1.50
]

# STEP 2: Test each threshold with and
# without the famous rural outlier.

membership_results = []

for threshold in membership_thresholds:

    for population_name, population in [
        ("With Munster", double_solitude),
        (
            "Without Munster",
            double_solitude[
                double_solitude["Address"]
                != "2890 Munster Rd"
            ]
        )
    ]:

        qualifies = (
            (population["Cart_to_Cart_km"] > threshold)
            &
            (population["Core_km"] > threshold)
        )

        parks = (
            population["Habitat_Group"]
            == "Park Maintenance"
        )

        other = ~parks

        park_members = int(
            (qualifies & parks).sum()
        )

        other_members = int(
            (qualifies & other).sum()
        )

        park_total = int(parks.sum())
        other_total = int(other.sum())

        park_rate = park_members / park_total
        other_rate = other_members / other_total

        membership_results.append({
            "Threshold_km": threshold,
            "Population": population_name,
            "Total_Members": int(qualifies.sum()),
            "Park_Members": park_members,
            "Park_Percent": park_rate * 100,
            "Other_Members": other_members,
            "Other_Percent": other_rate * 100,
            "Park_to_Other_Ratio": (
                park_rate / other_rate
                if other_rate > 0
                else np.nan
            )
        })

# STEP 3: Display the alternative club populations

club_threshold_summary = pd.DataFrame(
    membership_results
)

print("\nTHE FOUR MEMBERSHIP STANDARDS")

display(
    club_threshold_summary.round(2)
)

# STEP 4: Identify the members of the
# most exclusive 1.5 km club.

exclusive_members = double_solitude[
    (double_solitude["Cart_to_Cart_km"] > 1.5)
    &
    (double_solitude["Core_km"] > 1.5)
].copy()

print("\nTHE ULTRA-ISOLATED MEMBERS")
print("Both distances greater than 1.5 km")

display(
    exclusive_members[
        [
            "Address",
            "Habitat_Group",
            "Cart_to_Cart_km",
            "Core_km",
            "Nearest_Cart_Address"
        ]
    ].sort_values(
        "Cart_to_Cart_km",
        ascending=False
    ).round(2)
)

# STEP 5: Validate the original 1 km findings

original_rule = club_threshold_summary[
    club_threshold_summary["Threshold_km"] == 1.0
]

with_munster = original_rule[
    original_rule["Population"] == "With Munster"
].iloc[0]

without_munster = original_rule[
    original_rule["Population"] == "Without Munster"
].iloc[0]

assert with_munster["Total_Members"] == 9
assert with_munster["Park_Members"] == 3

assert without_munster["Total_Members"] == 8
assert without_munster["Park_Members"] == 2

assert len(club_threshold_summary) == 8

print("\n✓ Four membership thresholds tested.")
print("✓ Munster sensitivity examined.")
print("✓ Original nine-member club reconciled.")
print("✓ All source data preserved.")
```

```python

# ==========================================
# CHAPTER 8 — THE GREAT CART ESCAPE
# 05 | The Retailer Eligibility Appeals
# ==========================================

import pandas as pd
import numpy as np

print("THE RETAILER ELIGIBILITY APPEALS")
print("-" * 54)

# STEP 1: Bring forward the two retailer scenarios
# and the cart-to-cart isolation measurements.

appeals = double_solitude.copy()

assert len(appeals) == 214

# STEP 2: Test both retailer scenarios
# under the four membership thresholds.

thresholds = [0.75, 1.00, 1.25, 1.50]

appeal_results = []
lost_members = []

for threshold in thresholds:

    # Distance to nearest OTHER cart report
    far_from_carts = (
        appeals["Cart_to_Cart_km"] > threshold
    )

    # Original core-retailer eligibility
    core_members = (
        far_from_carts
        & (appeals["Core_km"] > threshold)
    )

    # Extended retailer eligibility
    extended_members = (
        far_from_carts
        & (appeals["Extended_km"] > threshold)
    )

    # The extended scenario contains all core stores.
    # Therefore, it cannot create NEW members.
    assert not (
        extended_members & ~core_members
    ).any()

    # A member loses eligibility if an additional
    # retailer brings the nearest-store distance
    # within the chosen threshold.
    lost = core_members & ~extended_members

    appeal_results.append({
        "Threshold_km": threshold,
        "Core_Members": int(core_members.sum()),
        "Extended_Members": int(
            extended_members.sum()
        ),
        "Membership_Lost": int(lost.sum()),
        "Membership_Retained": int(
            (core_members & extended_members).sum()
        ),
        "Park_Members_Extended": int(
            (
                extended_members
                & appeals["Park_Report"]
            ).sum()
        )
    })

    for _, member in appeals[lost].iterrows():

        lost_members.append({
            "Threshold_km": threshold,
            "Address": member["Address"],
            "Habitat_Group": member["Habitat_Group"],
            "Cart_to_Cart_km": member["Cart_to_Cart_km"],
            "Core_km": member["Core_km"],
            "Extended_km": member["Extended_km"],
            "New_Nearest_Store": member["Extended_Store"]
        })

# STEP 3: Display membership changes

appeal_summary = pd.DataFrame(appeal_results)

print("\nTHE FOUR APPEAL HEARINGS")

display(appeal_summary)

# STEP 4: Identify exactly who loses membership

lost_members_df = pd.DataFrame(
    lost_members,
    columns=[
        "Threshold_km",
        "Address",
        "Habitat_Group",
        "Cart_to_Cart_km",
        "Core_km",
        "Extended_km",
        "New_Nearest_Store"
    ]
)

print("\nMEMBERS WHO LOST ELIGIBILITY")
print("After admitting extended retailers")

if lost_members_df.empty:
    print("Nobody! All existing members retained.")
else:
    display(
        lost_members_df.round(2)
    )

# STEP 5: Inspect the executive members
# under BOTH retailer definitions.

executive_review = appeals[
    (appeals["Cart_to_Cart_km"] > 1.5)
    & (appeals["Core_km"] > 1.5)
].copy()

executive_review["Extended_Eligible"] = (
    executive_review["Extended_km"] > 1.5
)

print("\nTHE EXECUTIVE ISOLATION SOCIETY")
print("Do our three members survive the appeals?")

display(
    executive_review[
        [
            "Address",
            "Cart_to_Cart_km",
            "Core_km",
            "Extended_km",
            "Extended_Eligible"
        ]
    ].round(2)
)

# STEP 6: Validate the original membership counts

expected_core_counts = {
    0.75: 14,
    1.00: 9,
    1.25: 4,
    1.50: 3
}

for _, row in appeal_summary.iterrows():
    assert (
        row["Core_Members"]
        == expected_core_counts[row["Threshold_km"]]
    )

assert len(appeal_summary) == 4
assert len(executive_review) == 3

print("\n✓ Four retailer eligibility hearings completed.")
print("✓ Original core memberships reconciled.")
print("✓ Membership losses documented.")
print("✓ No new members created by adding retailers.")
print("✓ Source data preserved.")
```

```python

# ==========================================
# CHAPTER 8 — THE GREAT CART ESCAPE
# 06 | The Club's Weakest Link
# ==========================================

import pandas as pd
import numpy as np

print("THE CLUB'S WEAKEST LINK")
print("-" * 52)

# STEP 1: Bring forward the two isolation
# measurements for all 214 reporting locations.

isolation_scores = double_solitude.copy()

# STEP 2: Calculate the weaker (smaller)
# distance under each retailer scenario.
#
# A location must exceed BOTH distances to
# qualify for any given threshold.

isolation_scores["Core_Score_km"] = np.minimum(
    isolation_scores["Cart_to_Cart_km"],
    isolation_scores["Core_km"]
)

isolation_scores["Extended_Score_km"] = np.minimum(
    isolation_scores["Cart_to_Cart_km"],
    isolation_scores["Extended_km"]
)

# STEP 3: Identify which distance limits
# each location's isolation score.

for scenario, retailer_column in [
    ("Core", "Core_km"),
    ("Extended", "Extended_km")
]:

    isolation_scores[f"{scenario}_Limiting_Factor"] = (
        np.where(
            isolation_scores["Cart_to_Cart_km"]
            <= isolation_scores[retailer_column],
            "Nearest other cart report",
            "Nearest retailer candidate"
        )
    )

# STEP 4: See the strongest double-isolation
# locations WITHOUT choosing a threshold.

print("\nTHE DOUBLE ISOLATION LEADERBOARD")
print("Ranked by extended-retailer score")

display(
    isolation_scores.nlargest(
        15, "Extended_Score_km"
    )[
        [
            "Address",
            "Habitat_Group",
            "Cart_to_Cart_km",
            "Core_km",
            "Extended_km",
            "Extended_Score_km",
            "Extended_Limiting_Factor"
        ]
    ].round(2)
)

# STEP 5: Who is usually the weakest link?

print("\nWHAT LIMITS DOUBLE ISOLATION?")

bottleneck_summary = pd.crosstab(
    isolation_scores["Core_Limiting_Factor"],
    isolation_scores["Extended_Limiting_Factor"],
    margins=True
)

display(bottleneck_summary)

# STEP 6: Examine the executive society.
# Positive margin = remains eligible.
# Negative margin = does not qualify.

executive_scores = isolation_scores[
    isolation_scores["Core_Score_km"] > 1.5
].copy()

executive_scores["Executive_Margin_km"] = (
    executive_scores["Extended_Score_km"] - 1.5
)

print("\nEXECUTIVE MEMBERSHIP SECURITY")
print("Distance above the 1.5 km threshold")

display(
    executive_scores[
        [
            "Address",
            "Core_Score_km",
            "Extended_Score_km",
            "Executive_Margin_km",
            "Extended_Limiting_Factor"
        ]
    ].sort_values(
        "Executive_Margin_km",
        ascending=False
    ).round(3)
)

# STEP 7: Find applicants closest to the
# original 1 km club's admission boundary.

isolation_scores["Margin_1km"] = (
    isolation_scores["Extended_Score_km"] - 1.0
)

print("\nTHE 1 KM MEMBERSHIP BORDER")
print("Locations closest to qualifying or failing")

display(
    isolation_scores.assign(
        Distance_To_Border=lambda df:
        df["Margin_1km"].abs()
    ).nsmallest(
        12, "Distance_To_Border"
    )[
        [
            "Address",
            "Habitat_Group",
            "Extended_Score_km",
            "Margin_1km",
            "Extended_Limiting_Factor"
        ]
    ].round(3)
)

# STEP 8: Validate against previous hearings.

assert len(isolation_scores) == 214

assert (
    isolation_scores["Extended_Score_km"]
    <= isolation_scores["Core_Score_km"] + 1e-8
).all()

expected_members = {
    0.75: 13,
    1.00: 9,
    1.25: 4,
    1.50: 3
}

for threshold, expected in expected_members.items():

    actual = int(
        (
            isolation_scores["Extended_Score_km"]
            > threshold
        ).sum()
    )

    assert actual == expected, (
        f"Unexpected membership count at {threshold} km"
    )

assert len(executive_scores) == 3

print("\n✓ All 214 isolation scores calculated.")
print("✓ Membership thresholds reconciled.")
print("✓ Limiting distances identified.")
print("✓ No original data modified.")
```

```python

# ==========================================
# CHAPTER 9 — THE FOUR SEASONS OF CARTUS
# 05 | The Returning Residents
# ==========================================

import pandas as pd

print("THE RETURNING RESIDENTS")
print("-" * 54)

# STEP 1: Count reports at each address
# separately for all four seasons.
#
# IMPORTANT: These are reports, not
# confirmed sightings of individual carts.

seasonal_address_counts = pd.crosstab(
    seasonal_mapped["Address"],
    seasonal_mapped["Season"]
).reindex(
    columns=season_order,
    fill_value=0
)

# STEP 2: Count how many different seasons
# each address appears in.

seasonal_residents = seasonal_address_counts.copy()

seasonal_residents["Total_Reports"] = (
    seasonal_address_counts.sum(axis=1)
)

seasonal_residents["Seasons_Reported"] = (
    (seasonal_address_counts > 0).sum(axis=1)
)

# Create a readable seasonal signature.
# Example: "Spring, Summer, Fall"

seasonal_residents["Seasonal_Signature"] = (
    seasonal_address_counts.apply(
        lambda row: ", ".join(
            season
            for season in season_order
            if row[season] > 0
        ),
        axis=1
    )
)

# STEP 3: Preserve the first and last
# reported dates at each address.

date_ranges = (
    seasonal_mapped
    .groupby("Address")["Opened_Date"]
    .agg(
        First_Report="min",
        Last_Report="max"
    )
)

seasonal_residents = seasonal_residents.join(
    date_ranges
)

# STEP 4: Classify locations by reporting
# frequency and seasonal recurrence.

seasonal_residents["Multiple_Reports"] = (
    seasonal_residents["Total_Reports"] > 1
)

seasonal_residents["Multiple_Seasons"] = (
    seasonal_residents["Seasons_Reported"] > 1
)

print("\nTHE RESIDENCY REGISTER")

print(
    "Distinct reporting addresses:",
    len(seasonal_residents)
)

print(
    "Addresses with multiple reports:",
    int(seasonal_residents["Multiple_Reports"].sum())
)

print(
    "Addresses appearing in multiple seasons:",
    int(seasonal_residents["Multiple_Seasons"].sum())
)

# STEP 5: See how many locations were
# recorded in one, two, three or four seasons.

seasonal_residency_census = (
    seasonal_residents["Seasons_Reported"]
    .value_counts()
    .reindex([1, 2, 3, 4], fill_value=0)
    .rename_axis("Number_of_Seasons")
    .reset_index(name="Addresses")
)

print("\nTHE SEASONAL RESIDENCY CENSUS")
display(seasonal_residency_census)

# STEP 6: Find addresses with multiple
# reports that all occurred in ONE season.

same_season_repeaters = seasonal_residents[
    seasonal_residents["Multiple_Reports"]
    & ~seasonal_residents["Multiple_Seasons"]
].copy()

print("\nTHE SAME-SEASON REGULARS")

print(
    "Addresses with repeated reports "
    "but only one recorded season:",
    len(same_season_repeaters)
)

# STEP 7: Assemble the multi-season
# membership register.

returning_residents = seasonal_residents[
    seasonal_residents["Multiple_Seasons"]
].copy()

returning_residents = returning_residents.sort_values(
    by=[
        "Seasons_Reported",
        "Total_Reports",
        "First_Report"
    ],
    ascending=[False, False, True]
)

print("\nTHE OFFICIAL RETURNING RESIDENTS REGISTER")

display(
    returning_residents[
        [
            "Winter",
            "Spring",
            "Summer",
            "Fall",
            "Total_Reports",
            "Seasons_Reported",
            "Seasonal_Signature",
            "First_Report",
            "Last_Report"
        ]
    ].head(25)
)

# STEP 8: Quantify how much reporting
# activity occurred at multi-season places.

reports_at_returning_addresses = int(
    returning_residents["Total_Reports"].sum()
)

print("\nTHE COMMITTEE'S MEMBERSHIP REPORT")

print(
    "Reports from multi-season addresses:",
    reports_at_returning_addresses
)

print(
    "Share of all geocoded reports:",
    round(
        reports_at_returning_addresses
        / len(seasonal_mapped) * 100,
        1
    ),
    "%"
)

# STEP 9: Validate against the previous work.

assert len(seasonal_mapped) == 247
assert len(seasonal_residents) == 214

assert (
    seasonal_residents["Total_Reports"].sum()
    == 247
)

assert (
    247 - len(seasonal_residents)
    == 33
)

assert (
    seasonal_residency_census["Addresses"].sum()
    == 214
)

assert (
    len(returning_residents)
    <= seasonal_residents["Multiple_Reports"].sum()
)

print("\n✓ All 247 geocoded reports accounted for.")
print("✓ All 214 distinct addresses reconciled.")
print("✓ Seasonal appearances counted independently.")
print("✓ Original records unchanged.")
```

```python

# ==========================================
# CHAPTER 9 — THE FOUR SEASONS OF CARTUS
# 06 | Field Map No. 006
# THE SEASONAL MIGRATION ATLAS
# ==========================================

import pandas as pd
import folium
from branca.element import Element
from html import escape

print("THE SEASONAL MIGRATION ATLAS")
print("-" * 54)

# STEP 1: Prepare the original geocoded reports.
# No cart reports will be deleted or invented.

atlas_reports = seasonal_mapped.copy()

# Identify the existing coordinate columns.

column_lookup = {
    c.lower(): c for c in atlas_reports.columns
}

latitude_col = next(
    (
        column_lookup[c]
        for c in ["latitude", "lat"]
        if c in column_lookup
    ),
    None
)

longitude_col = next(
    (
        column_lookup[c]
        for c in ["longitude", "lon", "lng", "long"]
        if c in column_lookup
    ),
    None
)

if latitude_col is None or longitude_col is None:
    raise KeyError(
        "Could not locate latitude/longitude columns. "
        f"Available columns: {list(atlas_reports.columns)}"
    )

atlas_reports[latitude_col] = pd.to_numeric(
    atlas_reports[latitude_col],
    errors="coerce"
)

atlas_reports[longitude_col] = pd.to_numeric(
    atlas_reports[longitude_col],
    errors="coerce"
)

assert atlas_reports[
    [latitude_col, longitude_col]
].notna().all().all()

atlas_reports["Report_Day"] = (
    atlas_reports["Opened_Date"]
    .dt.strftime("%Y-%m-%d")
)

# STEP 2: Combine repeated reports occurring
# at the same address IN THE SAME SEASON.
#
# The underlying 247 reports stay untouched.

atlas_points = (
    atlas_reports
    .groupby(["Address", "Season"], as_index=False)
    .agg(
        Reports=("Season", "size"),
        Latitude=(latitude_col, "first"),
        Longitude=(longitude_col, "first"),
        Report_Dates=(
            "Report_Day",
            lambda x: ", ".join(sorted(x))
        ),
        Service_Families=(
            "Service_Family",
            lambda x: ", ".join(
                sorted(set(x.dropna().astype(str)))
            )
        )
    )
)

# STEP 3: Find the multi-season addresses.

multi_season_addresses = set(
    seasonal_residents[
        seasonal_residents["Seasons_Reported"] > 1
    ].index
)

# STEP 4: Assign the four official colours.

season_colours = {
    "Winter": "#4f9ed8",
    "Spring": "#2ca25f",
    "Summer": "#edaa32",
    "Fall": "#9867c5"
}

season_symbols = {
    "Winter": "❄️",
    "Spring": "🌷",
    "Summer": "☀️",
    "Fall": "🍂"
}

# STEP 5: Initialize the map.
# We keep the reliable ESRI basemap.

seasonal_map = folium.Map(
    location=[
        atlas_points["Latitude"].mean(),
        atlas_points["Longitude"].mean()
    ],
    zoom_start=10,
    tiles=None,
    control_scale=True,
    max_zoom=16
)

folium.TileLayer(
    tiles=(
        "https://server.arcgisonline.com/ArcGIS/rest/"
        "services/Canvas/World_Light_Gray_Base/"
        "MapServer/tile/{z}/{y}/{x}"
    ),
    attr=(
        "Tiles © Esri, HERE, Garmin, "
        "OpenStreetMap contributors, GIS user community"
    ),
    name="Esri Light Gray",
    max_zoom=16
).add_to(seasonal_map)

# STEP 6: Build four independent season layers.

season_layers = {}

for season in season_order:

    report_total = int(
        (atlas_reports["Season"] == season).sum()
    )

    layer = folium.FeatureGroup(
        name=(
            f"{season_symbols[season]} "
            f"{season} — {report_total} reports"
        ),
        show=True
    )

    subset = atlas_points[
        atlas_points["Season"] == season
    ]

    for _, point in subset.iterrows():

        address = escape(str(point["Address"]))

        dates = escape(str(point["Report_Dates"]))

        services = escape(
            str(point["Service_Families"])
        )

        report_count = int(point["Reports"])

        # Larger markers indicate multiple reports
        # at this address during this season.

        marker_radius = min(
            5.5 + (report_count - 1) * 1.5,
            11
        )

        popup_html = f"""
        <div style="
            font-family:Arial,sans-serif;
            min-width:220px;
            line-height:1.6;
        ">
            <b>{season_symbols[season]}
            {address}</b><br>
            <b>Season:</b> {season}<br>
            <b>Reports this season:</b>
            {report_count}<br>
            <b>Reported dates:</b>
            {dates}<br>
            <b>Service family:</b>
            {services}<br>
            <hr>
            <small>
            Reporting location, not a confirmed
            individual cart or travel route.
            </small>
        </div>
        """

        folium.CircleMarker(
            location=[
                point["Latitude"],
                point["Longitude"]
            ],
            radius=marker_radius,
            color="white",
            weight=1.2,
            fill=True,
            fill_color=season_colours[season],
            fill_opacity=0.88,
            tooltip=(
                f"{address} | {season} | "
                f"{report_count} report(s)"
            ),
            popup=folium.Popup(
                popup_html,
                max_width=340
            )
        ).add_to(layer)

    layer.add_to(seasonal_map)
    season_layers[season] = layer

# STEP 7: Add gold rings to addresses
# with reports in multiple seasons.
#
# These rings are NOT extra reports.
# They are a visual annotation only.

returning_layer = folium.FeatureGroup(
    name="🏡 Multi-season addresses (10)",
    show=True
)

for address in sorted(multi_season_addresses):

    address_rows = atlas_points[
        atlas_points["Address"] == address
    ]

    first_point = address_rows.iloc[0]

    membership = seasonal_residents.loc[address]

    folium.CircleMarker(
        location=[
            first_point["Latitude"],
            first_point["Longitude"]
        ],
        radius=11,
        color="#b8860b",
        weight=3,
        fill=False,
        opacity=0.95,
        tooltip=(
            f"Multi-season reporting address: "
            f"{escape(str(address))}"
        ),
        popup=folium.Popup(
            (
                f"<b>🏡 Multi-season address</b><br>"
                f"{escape(str(address))}<br>"
                f"Seasons: "
                f"{escape(str(membership['Seasonal_Signature']))}"
                f"<br>Total reports: "
                f"{int(membership['Total_Reports'])}"
            ),
            max_width=300
        )
    ).add_to(returning_layer)

returning_layer.add_to(seasonal_map)

# STEP 8: Add independent layer controls.

folium.LayerControl(
    collapsed=False
).add_to(seasonal_map)

# STEP 9: Provide the peer-reviewed legend.

seasonal_legend = """
<div style="
    position:fixed;
    bottom:35px;
    right:20px;
    z-index:9999;
    background:rgba(255,255,255,0.97);
    border:1px solid #d1d5db;
    border-radius:8px;
    padding:12px 14px;
    font-family:Arial,sans-serif;
    font-size:12px;
    line-height:1.8;
    color:#263238;
    box-shadow:0 2px 8px rgba(0,0,0,.12);
    pointer-events:none;
">

<b>🛒 The Seasonal Migration Atlas</b>

<div>
<span style="color:#4f9ed8;">●</span>
Winter — 29 reports
</div>

<div>
<span style="color:#2ca25f;">●</span>
Spring — 85 reports
</div>

<div>
<span style="color:#edaa32;">●</span>
Summer — 82 reports
</div>

<div>
<span style="color:#9867c5;">●</span>
Fall — 51 reports
</div>

<div style="
    border-top:1px solid #e5e7eb;
    margin-top:5px;
    padding-top:5px;
">
<span style="color:#b8860b;">◯</span>
Gold ring = multi-season address
</div>

<div style="font-size:10px;color:#64748b;
            line-height:1.4;margin-top:5px;">
    Larger dot = more reports in that season.<br>
    247 reports represented by 224 markers.<br>
    Reporting dates ≠ cart arrival dates.<br>
    Toggle seasons to see overlapping locations.
</div>
</div>
"""

seasonal_map.get_root().html.add_child(
    Element(seasonal_legend)
)

# STEP 10: Fit the map to all reported locations.

seasonal_map.fit_bounds([
    [
        atlas_points["Latitude"].min(),
        atlas_points["Longitude"].min()
    ],
    [
        atlas_points["Latitude"].max(),
        atlas_points["Longitude"].max()
    ]
])

# STEP 11: Validate everything.

assert len(atlas_reports) == 247
assert len(atlas_points) == 224

assert atlas_points["Reports"].sum() == 247

assert len(multi_season_addresses) == 10

assert len(season_layers) == 4

print("\nSEASONAL ATLAS INSPECTION")
print("✓ Esri Light Gray basemap configured.")
print("✓ Four independently selectable seasons.")
print("✓ 247 reports represented by 224 markers.")
print("✓ Ten multi-season addresses ringed in gold.")
print("✓ Repeated same-season reports preserved.")
print("✓ Original report records unchanged.")
print("✓ Straight-line coordinates; no travel paths.")

display(seasonal_map)
```

```python

# ==========================================
# CHAPTER 9 — THE FOUR SEASONS OF CARTUS
# 07 | The Residency Permit Audit
# ==========================================

import pandas as pd
from itertools import combinations

print("THE RESIDENCY PERMIT AUDIT")
print("-" * 54)

# STEP 1: Examine only the ten addresses
# with reports in more than one season.

permit_addresses = set(returning_residents.index)

permit_reports = seasonal_mapped[
    seasonal_mapped["Address"].isin(permit_addresses)
].copy()

# STEP 2: For each address, compare reporting
# dates belonging to DIFFERENT seasons.
#
# Find the shortest calendar-day separation.

permit_records = []

for address, group in permit_reports.groupby("Address"):

    records = list(
        group[
            ["Opened_Date", "Season"]
        ]
        .dropna()
        .sort_values("Opened_Date")
        .itertuples(index=False, name=None)
    )

    cross_season_pairs = []

    for first, second in combinations(records, 2):

        date_a, season_a = first
        date_b, season_b = second

        if season_a == season_b:
            continue

        gap_days = abs(
            (date_b.date() - date_a.date()).days
        )

        cross_season_pairs.append(
            (
                gap_days,
                date_a,
                season_a,
                date_b,
                season_b
            )
        )

    if not cross_season_pairs:
        raise ValueError(
            f"No cross-season pair found for {address}"
        )

    shortest = min(
        cross_season_pairs,
        key=lambda pair: pair[0]
    )

    gap, date_a, season_a, date_b, season_b = shortest

    permit_records.append({
        "Address": address,
        "Total_Reports": len(records),
        "Seasonal_Signature":
            seasonal_residents.loc[
                address, "Seasonal_Signature"
            ],
        "First_Season": season_a,
        "First_Date": date_a.date(),
        "Second_Season": season_b,
        "Second_Date": date_b.date(),
        "Shortest_Cross_Season_Gap_Days": gap,
        "Within_21_Days": gap <= 21
    })

# STEP 3: Build the official hearing register.

residency_permits = pd.DataFrame(
    permit_records
).sort_values(
    "Shortest_Cross_Season_Gap_Days"
)

print("\nTHE TEN RESIDENCY APPLICATIONS")

display(residency_permits)

# STEP 4: Separate quick calendar-boundary
# crossings from longer-separated reports.

quick_crossings = int(
    residency_permits["Within_21_Days"].sum()
)

longer_separations = int(
    (~residency_permits["Within_21_Days"]).sum()
)

print("\nTHE MEMBERSHIP COMMITTEE'S FINDINGS")

print("Multi-season addresses:",
      len(residency_permits))

print("Shortest cross-season gap <= 21 days:",
      quick_crossings)

print("Shortest cross-season gap > 21 days:",
      longer_separations)

print("\nIMPORTANT:")
print(
    "This is a calendar-gap sensitivity check, "
    "not an individual-cart identification "
    "or a replacement for episode clustering."
)

# STEP 5: Validate the original populations.

assert len(residency_permits) == 10
assert len(permit_reports) == 25

assert (
    quick_crossings + longer_separations
) == 10

assert (
    residency_permits[
        "Shortest_Cross_Season_Gap_Days"
    ] >= 0
).all()

print("\n✓ All ten multi-season addresses reviewed.")
print("✓ All 25 associated reports retained.")
print("✓ Actual calendar-day gaps measured.")
print("✓ Original records unchanged.")
```

```python

# ==========================================
# CHAPTER 9 — THE FOUR SEASONS OF CARTUS
# 12 | The Woodroffe Witness Lineup
# ==========================================

import pandas as pd
import numpy as np
from itertools import combinations

print("THE WOODROFFE WITNESS LINEUP")
print("-" * 54)

# STEP 1: Identify the three witnesses.
# No assumptions about geographical proximity.

witness_addresses = [
    "1701 Woodroffe Ave",
    "1645 Woodroffe Ave",
    "131 Greenbank Rd"
]

# STEP 2: Retrieve their original
# April-May Park Maintenance reports.

witness_reports = spring_rush[
    spring_rush["Address"].isin(witness_addresses)
].copy()

witness_reports["Report_Day"] = (
    pd.to_datetime(witness_reports["Opened_Date"])
    .dt.normalize()
)

# STEP 3: Retrieve the original locations
# from the established habitat table.

witness_locations = (
    habitat_points[
        habitat_points["Address"].isin(
            witness_addresses
        )
    ][["Address", "Latitude", "Longitude"]]
    .drop_duplicates("Address")
    .set_index("Address")
)

assert len(witness_locations) == 3

# STEP 4: Summarize reporting at each address.

witness_summary = (
    witness_reports
    .groupby("Address")
    .agg(
        Reports=("Address", "size"),
        Reporting_Days=("Report_Day", "nunique"),
        First_Report=("Report_Day", "min"),
        Last_Report=("Report_Day", "max")
    )
)

# Bring in our episode recount.

witness_episode_counts = (
    park_episode_summary[
        park_episode_summary["Address"].isin(
            witness_addresses
        )
    ]
    .groupby("Address")
    .size()
    .rename("Estimated_Episodes")
)

witness_summary = (
    witness_summary
    .join(witness_episode_counts)
    .join(witness_locations)
    .reindex(witness_addresses)
)

print("\nTHE THREE WITNESSES")
display(witness_summary)

# STEP 5: Compare actual reporting days.
# This is useful for spotting coinciding dates.

witness_calendar = (
    witness_reports
    .groupby(["Report_Day", "Address"])
    .size()
    .unstack(fill_value=0)
    .reindex(
        columns=witness_addresses,
        fill_value=0
    )
)

witness_calendar["Total_Reports"] = (
    witness_calendar.sum(axis=1)
)

print("\nTHE WITNESS ATTENDANCE CALENDAR")
display(witness_calendar)

# STEP 6: Calculate straight-line separation.
# Same Haversine principle as earlier chapters.

def witness_haversine(lat1, lon1, lat2, lon2):

    lat1, lon1, lat2, lon2 = np.radians(
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 6371.0088 * (
        2 * np.arcsin(np.sqrt(a))
    )

distance_records = []

for address_a, address_b in combinations(
    witness_addresses, 2
):

    a = witness_locations.loc[address_a]
    b = witness_locations.loc[address_b]

    distance = witness_haversine(
        float(a["Latitude"]),
        float(a["Longitude"]),
        float(b["Latitude"]),
        float(b["Longitude"])
    )

    distance_records.append({
        "First_Address": address_a,
        "Second_Address": address_b,
        "Straight_Line_km": round(distance, 3)
    })

witness_distances = pd.DataFrame(
    distance_records
).sort_values("Straight_Line_km")

print("\nHOW CLOSE ARE THE WITNESSES?")
display(witness_distances)

# STEP 7: Reconcile all reported activity.

assert len(witness_reports) == 10
assert witness_summary["Reports"].sum() == 10
assert witness_summary["Estimated_Episodes"].sum() == 3
assert len(witness_distances) == 3

assert witness_calendar["Total_Reports"].sum() == 10

print("\nTHE COMMITTEE'S PRELIMINARY FINDINGS")
print("-" * 54)

print("Addresses investigated:", len(witness_addresses))
print("Original reports:", len(witness_reports))
print("Estimated episodes:",
      int(witness_summary["Estimated_Episodes"].sum()))
print("Distinct reporting dates:",
      len(witness_calendar))

print("\n✓ All ten reports accounted for.")
print("✓ Three actual reporting locations compared.")
print("✓ Geographic distances measured.")
print("✓ Same-day reporting dates preserved.")
print("✓ No locations or reports merged.")
print("✓ Original records unchanged.")
```

```python

# ==========================================
# CHAPTER 9 — THE FOUR SEASONS OF CARTUS
# 13 | The Neighbourhood Conspiracy
# ==========================================

import pandas as pd
import numpy as np
from itertools import combinations

print("THE NEIGHBOURHOOD CONSPIRACY")
print("-" * 54)

# STEP 1: Use the existing Park Maintenance
# episode estimates from Investigation 11.

neighbourhood_episodes = (
    park_episode_summary.copy()
)

# STEP 2: Attach the established geographical
# coordinates for each reporting address.

park_coordinates = (
    habitat_points[
        ["Address", "Latitude", "Longitude"]
    ]
    .drop_duplicates("Address")
)

neighbourhood_episodes = (
    neighbourhood_episodes.merge(
        park_coordinates,
        on="Address",
        how="left",
        validate="many_to_one"
    )
)

assert neighbourhood_episodes[
    ["Latitude", "Longitude", "First_Report"]
].notna().all().all()

# STEP 3: Establish the exploratory rules.

distance_limit_m = 300
time_limit_days = 21

# I am comparing independent addresses,
# NOT combining their original reports.

def neighbourhood_distance_m(a, b):

    lat1, lon1 = np.radians([
        float(a["Latitude"]),
        float(a["Longitude"])
    ])

    lat2, lon2 = np.radians([
        float(b["Latitude"]),
        float(b["Longitude"])
    ])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    h = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 6371008.8 * 2 * np.arcsin(
        np.sqrt(np.clip(h, 0, 1))
    )

# STEP 4: Compare each pair of episodes
# recorded at DIFFERENT addresses.

pair_records = []

episodes = list(
    neighbourhood_episodes.to_dict("records")
)

for a, b in combinations(episodes, 2):

    if a["Address"] == b["Address"]:
        continue

    separation_m = neighbourhood_distance_m(a, b)

    gap_days = abs(
        (
            pd.Timestamp(a["First_Report"]).normalize()
            - pd.Timestamp(b["First_Report"]).normalize()
        ).days
    )

    close_space = separation_m <= distance_limit_m
    close_time = gap_days <= time_limit_days

    pair_records.append({
        "Address_A": a["Address"],
        "Address_B": b["Address"],
        "Distance_m": round(separation_m, 1),
        "Episode_Start_A": pd.Timestamp(
            a["First_Report"]
        ).date(),
        "Episode_Start_B": pd.Timestamp(
            b["First_Report"]
        ).date(),
        "Start_Gap_Days": gap_days,
        "Within_300m": close_space,
        "Within_21_Days": close_time,
        "Space_Time_Candidate": (
            close_space and close_time
        )
    })

park_neighbourhood_pairs = pd.DataFrame(
    pair_records
)

# STEP 5: Inspect the geographically
# closest pairs, regardless of time.

print("\nTHE TEN CLOSEST PARK EPISODE PAIRS")

display(
    park_neighbourhood_pairs
    .sort_values("Distance_m")
    .head(10)
)

# STEP 6: Identify pairs that satisfy BOTH
# of the exploratory conditions.

neighbourhood_candidates = (
    park_neighbourhood_pairs[
        park_neighbourhood_pairs[
            "Space_Time_Candidate"
        ]
    ]
    .sort_values(
        ["Distance_m", "Start_Gap_Days"]
    )
)

print("\nTHE SPACE–TIME WITNESS REGISTER")

print(
    "Pairs within 300m AND 21 days:",
    len(neighbourhood_candidates)
)

if len(neighbourhood_candidates):

    display(
        neighbourhood_candidates[
            [
                "Address_A",
                "Address_B",
                "Distance_m",
                "Episode_Start_A",
                "Episode_Start_B",
                "Start_Gap_Days"
            ]
        ]
    )

else:
    print(
        "No pairs met both conditions."
    )

# STEP 7: Check whether the Woodroffe
# witnesses appear on the candidate list.

woodroffe_pair = neighbourhood_candidates[
    (
        (
            neighbourhood_candidates["Address_A"]
            == "1701 Woodroffe Ave"
        )
        &
        (
            neighbourhood_candidates["Address_B"]
            == "1645 Woodroffe Ave"
        )
    )
    |
    (
        (
            neighbourhood_candidates["Address_A"]
            == "1645 Woodroffe Ave"
        )
        &
        (
            neighbourhood_candidates["Address_B"]
            == "1701 Woodroffe Ave"
        )
    )
]

print("\nTHE WOODROFFE MEMBERSHIP CHECK")

print(
    "Woodroffe pair identified:",
    not woodroffe_pair.empty
)

if not woodroffe_pair.empty:
    display(
        woodroffe_pair[
            [
                "Address_A",
                "Address_B",
                "Distance_m",
                "Start_Gap_Days"
            ]
        ]
    )

# STEP 8: Validate the research population.

assert len(park_episode_summary) == 29

assert len(neighbourhood_episodes) == 29

assert len(park_neighbourhood_pairs) == 406

assert not woodroffe_pair.empty

assert (
    neighbourhood_candidates["Within_300m"]
).all()

assert (
    neighbourhood_candidates["Within_21_Days"]
).all()

print("\nTHE COMMITTEE'S INSPECTION")
print("-" * 54)

print("✓ All 29 Park episodes preserved.")
print("✓ All 406 address pairs compared.")
print("✓ Distance and date criteria tested.")
print("✓ Woodroffe witnesses identified.")
print("✓ No reports or episodes merged.")
print("✓ Original records unchanged.")
```

### The Sensitivity Hearing

When both Woodroffe addresses were temporarily excluded, the April–May share of annual Park Maintenance reports decreased from **61.1% to 50.0%**.

The corresponding share of estimated reporting episodes decreased from **51.7% to 48.1%**.

However, the full March–May season still accounted for **53.6% of remaining reports** and **51.9% of remaining estimated episodes**.

The evidence therefore suggested that the Woodroffe locations amplified the observed spring concentration, while other reporting locations also contributed to it.

### Scientific Qualifications

This investigation examines municipal service requests, not verified individual shopping carts.

The missing-coordinate population is substantial and unevenly distributed across service families and seasons.

Estimated reporting episodes depend on an exploratory 21-day heuristic. Proximity in space and time does not establish a shared cart, cause, route, or incident.

The analysis covers one calendar year and cannot establish recurring annual seasonality.

The observed spring concentration could reflect reporting practices, maintenance activity, public awareness, underlying cart presence, or a combination of factors. These explanations were not independently verified.

### Final Field Observation

**Seasonal reporting patterns are shaped by more than the dates on which complaints are recorded.**

Coordinate availability, service classifications, repeated reporting at particular addresses, and the choice between report counts and estimated episodes all materially affect interpretation.

A visible spring concentration remained after sensitivity testing, but its magnitude depended on the counting method and inclusion of influential reporting locations.

The committee has therefore classified the phenomenon as:

**The Spring Awakening — A Documented Seasonal Reporting Concentration of Unconfirmed Origin.**
